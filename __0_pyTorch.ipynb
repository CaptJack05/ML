Two main rules that permfomrin matrix multiplication need to satisfy

### Reshaping , stacking , sqeezing and unsqueezing tensors

* Reshaping - reshapes an input tnesor to a defined shape
* View - Return a view of an input tensor of certain shpae but keep the same memory as original tensor
* Stacking - Combine mutiple tensors on top of each other (vstack) or side by side(hstack)
* Sqeeze- remove all 1 dimension from a tensor
* Unsqeeze - add a 1 dimension to a target tensor
* permute - return a view of the input with dimension permuted(swapped) in a certain way

```python
x = torch.arange(1,10)
x, x.shape
```

```python
#add an extra dimension
x_reshaped = x.reshape(3,3)
x_reshaped, x_reshaped.shape
```

```python
x
```

```python
#Change the view
z = x.view(1,9)
z, z.shape
```

```python
#changing z changes x as a view of tensor shares the same memory as the original tensor

z[:,0] = 5
z,x
```
