### Random Tesors


Why random tesors ?

Random tesnosors are important because the way many neural ntworks learns is that tehy start with tensors full of random numbers and then adjust those random numvers to better represent the data.

'Start with random numbers -> look at data -> update random numbers -> look at data -> update random nmbers '

```python
# Create a random tensor

random_tensor = torch.rand(1,3,10)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor.shape
```

```python
#Random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(3,224,224))
random_image_size_tensor
```

```python
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

### Tenosr Datatype

**NOTE:** Tesnosr datatype is one of the three vig errors with PyTorch & Deeplearning:

1. Tensor not right datatype
2. Tenosr not right shape
3. Tenosr not on the right device

```python
#float 32 tensor
float_32_tesnor = torch.tensor([3.0,6.0,9.0],
                               dtype= None,             # What datatype is the tenosr (float32 float 16 etc)
                               device = None,            # what device is your tenosr on
                               requires_grad = False)     # Whether or not to track gradiens with this tensors operation
float_32_tesnor
```

```python
# float 16 tesnsor
float_16_tensor = float_32_tesnor.type(torch.float16)
float_16_tensor
```

```python
float_16_tensor*float_32_tesnor
```

### Getting information from tensors (attributes)

1. Tensor not right datatype -> tensor.dtype
2. Tenosr not right shape -> tensor.shape
3. Tenosr not on the right device -> tensor.device

```python
#Random Tensor

some_tensor = torch.rand(3,4)
some_tensor
```

```python
#finding details of tensor

print(some_tensor)
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device of tensor: {some_tensor.device}")
```
